```python
df['systolic_bp'].isna().sum()
```

```python
df['systolic_bp'] = df['systolic_bp'].fillna(mean_sys_bp)
```

Transform variables

```python
def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "Underweight"
  elif bmi < 25:
    bmi_result = "Normal"
  elif bmi < 30:
    bmi_result = "Overweight"
  else:
    bmi_result = "Obese"
  return bmi_result
```

```python
df['bmi_category'] = df['bmi'].apply(bmi_check)

df['bmi_category'].value_counts()
```

```python
df['bmi_category'] = pd.cut(df['bmi'],
                            bins=[0, 18.5, 25, 30, 100],
                            labels=['Underweight', 'Normal', 'Overweight', 'Obese'])

# dropna=False also counts the missing values, so we can see them
df['bmi_category'].value_counts(dropna=False)
```

```python
df['education'].unique()
```

```python
df['education'].head()
```

```python
# Create a mapping dictionary (adjust based on your data dictionary)
education_map = {
    1.0: 'Less than 9th grade',
    2.0: '9-11th grade',
    3.0: 'High school grad',
    4.0: 'Some college or AA degree',
    5.0: 'College graduate',
    9.0: 'Do not know',
    'None': 'Missing'
}
# Apply the mapping
df['education_label'] = df['education'].map(education_map)
```

```python
df['education_label'].value_counts(dropna = False)
```
