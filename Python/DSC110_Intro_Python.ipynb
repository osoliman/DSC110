# **Functions**

**Temperature**

```python
def celsius_to_fahrenheit(celsius):
  fahrenheit = (celsius * 9/5) + 32
  return fahrenheit
```

```python
celsius_to_fahrenheit(0)
```

IF statements

```python
temp = 10

if temp < 0:
  print('It is cold')
else:
    print('It is not cold')
```

Formatted String Literals

```python
temp = 10

if temp < 0:
  print(f"temperature of {temp}F means it is cold")
else:
    print(f"temperature of {temp}F means it is not cold")
```

# **Loops**

```python
Celsius_temps_to_convert = [-20, 0, 10, 20, 30, 40]
```

```python
for temp in Celsius_temps_to_convert:
  print(celsius_to_fahrenheit(temp))
```

```python
for temp in Celsius_temps_to_convert:
  if celsius_to_fahrenheit(temp) <32:
    print(f"temperature of {celsius_to_fahrenheit(temp)}F means it is cold")
  else:
    print(f"temperature of {celsius_to_fahrenheit(temp)}F means it is not cold")
```
