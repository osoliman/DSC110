# Functions

```python
#Creating the function
def add_one(number):
  return number + 1
```

```python
#Using the function or calling the function
add_one(3)
```

```python
def bp_check(systolic_bp):
  if systolic_bp > 120:
    bp_result = "High"
  else:
    bp_result = "Normal"
  return bp_result
```

```python
bp_check(150)
```

# Loops

```python
bp_list = [120, 135, 170, 110, 100]
```

```python
for bp in bp_list:
  print(bp_check(bp))
```

```python
def bmi_check(bmi):
  if bmi < 18:
    print("Underweight")
  if bmi > 20 < 25:
    print("Normal")
  if bmi > 25 < 30:
    print("Obese")
  if bmi > 35:
    print("Overweight")
  return bmi
```

```python
def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "Underweight"
  elif bmi < 25:
    bmi_result = "Normal"
  elif bmi < 30:
    bmi_result = "Overweight"
  else:
    bmi_result = "Obese"
  return bmi_result
```

```python
bmi_check(24.9)
```
