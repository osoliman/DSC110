Conditionals

```python
my_systolic_bp = 140

if my_systolic_bp > 120:
  print('High')
else:
  print('Normal')
```

```python
def bp_check(bp_reading):
    if bp_reading > 120:
        bp_result = 'high'
    else:
        bp_result = 'normal'
    return bp_result
```

```python
print(bp_check(135))
```

```python
print(bp_check(110))
```

Loops

```python
bp_values = [120,130,150,90]
for bp in bp_values:
  print(bp_check(bp))
```
